```python
import tkinter as tk
from gtts import gTTS
from deep_translator import GoogleTranslator
import os

# ایجاد پنجره اصلی
root = tk.Tk()
root.title("ترجمه و صدا")
root.geometry("600x400")
root.config(bg="#3e2b29")  # رنگ پس زمینه قهوه ای

# متن خوشامدگویی
welcome_label = tk.Label(root, text="خوش آمدید!", font=("Arial", 20), fg="white", bg="#3e2b29")
welcome_label.pack(pady=20)

# ورودی برای جمله
entry_label = tk.Label(root, text="یک جمله به من بده:", font=("Arial", 14), fg="white", bg="#3e2b29")
entry_label.pack(pady=10)

entry = tk.Entry(root, font=("Arial", 14), width=30)
entry.pack(pady=10)

# تابع برای ترجمه و پخش صدا
def translate_and_play():
    user_input = entry.get()
    if user_input:
        # ترجمه به انگلیسی
        translated_text = GoogleTranslator(source='auto', target='en').translate(user_input)

        # تبدیل متن به صدا به زبان انگلیسی
        voice = gTTS(text=translated_text, lang='en')  # تغییر زبان به انگلیسی
        voice.save("voice.mp3")
        os.system("start voice.mp3")  # پخش صدا در ویندوز

# دکمه برای شروع ترجمه و پخش صدا
translate_button = tk.Button(root, text="ترجمه و پخش صدا", font=("Arial", 14), fg="white", bg="#f39c12", command=translate_and_play)
translate_button.pack(pady=20)

# اضافه کردن پس‌زمینه ستارگان
canvas = tk.Canvas(root, width=600, height=400, bg="#3e2b29", bd=0, highlightthickness=0)
canvas.pack()

# ستارگان متحرک
import random

def create_star():
    x = random.randint(0, 600)
    y = random.randint(0, 400)
    size = random.randint(2, 5)
    canvas.create_oval(x, y, x+size, y+size, fill="white", outline="white")

# ایجاد ستارگان به صورت تصادفی
for _ in range(50):
    create_star()

# اجرای پنجره
root.mainloop()
```
